## Topic 1 – What are Interaction Features?

### Step 1 – Understand the Concept

Interaction Features are new features created by combining two or more existing features.

They help a Machine Learning model understand how one feature influences another feature.

Sometimes, two features together provide more useful information than considering them separately.

For example:
- Income and Age together may help understand customer purchasing behaviour.
- Quantity and Price together help calculate the total purchase amount.
- Experience and Age together may help understand career experience relative to age.

Interaction features help Machine Learning models identify relationships between different variables.


### Step 2 – Demonstrate the Concept

Real-world example:

Consider an online shopping company.

The company has the following customer information:

| Customer | Income | Age |
|---|---:|---:|
| A | 30000 | 22 |
| B | 50000 | 30 |
| C | 80000 | 40 |

The company wants to understand customer purchasing behaviour.

Instead of analysing Income and Age separately, it can create a new feature by multiplying them.

Business use:
- Understand customer purchasing patterns.
- Identify relationships between customer attributes.
- Improve customer segmentation.

AI/ML use:
Interaction features can help a Machine Learning model identify combined patterns that may influence its predictions.


### Step 3 – Implement the Concept

We will create a simple dataset containing customer income and age.

Then, we will create a new interaction feature by combining both columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C"],
    "Income": [30000, 50000, 80000],
    "Age": [22, 30, 40]
})

df["Income_Age_Interaction"] = df["Income"] * df["Age"]

print(df)
```

### Output Explanation

The dataset contains the original Income and Age features.

The new Income_Age_Interaction column is created by multiplying Income by Age.

For example:

Customer A:
30000 × 22 = 660000

The new column represents the combined information from both features.

This is an example of an interaction feature.

## Topic 2 – Feature × Feature

### Step 1 – Understand the Concept

Feature × Feature means multiplying two existing features to create a new feature.

Formula:

New Feature = Feature 1 × Feature 2

This operation helps capture combined relationships between two numerical variables.

Examples:
- Income × Age
- Quantity × Price
- Marketing Spend × Website Traffic

The resulting feature may help a Machine Learning model understand patterns that are not obvious from the original features.



### Step 2 – Demonstrate the Concept

Real-world example:

An online shopping company records the quantity of products purchased and the price of each product.

| Product | Quantity | Price |
|---|---:|---:|
| Laptop | 2 | 40000 |
| Mobile | 3 | 15000 |
| Headphones | 4 | 2000 |

The company can multiply Quantity by Price to calculate the purchase amount.

Business use:
- Calculate total product value.
- Understand customer spending.
- Analyse product revenue.

AI/ML use:
The combined feature can help models identify relationships between purchase quantity and price.


### Step 3 – Implement the Concept

Create a Quantity × Price interaction feature.

The result represents the total value of each product purchase.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones"],
    "Quantity": [2, 3, 4],
    "Price": [40000, 15000, 2000]
})

df["Quantity_Price"] = df["Quantity"] * df["Price"]

print(df)
```

### Output Explanation

The Quantity_Price column is created by multiplying Quantity and Price.

Examples:

Laptop: 2 × 40000 = 80000

Mobile: 3 × 15000 = 45000

Headphones: 4 × 2000 = 8000

This feature represents the total purchase value for each product.

## Topic 3 – Feature + Feature

### Step 1 – Understand the Concept

Feature + Feature means adding two existing numerical features to create a new feature.

Formula:

New Feature = Feature 1 + Feature 2

Addition is useful when two values contribute to a combined total.

Examples:
- Basic Salary + Bonus
- Online Spending + Offline Spending
- Product A Sales + Product B Sales

The resulting feature represents the combined value of both features.



### Step 2 – Demonstrate the Concept

Real-world example:

An employee receives a basic salary and a performance bonus.

| Employee | Basic Salary | Bonus |
|---|---:|---:|
| A | 25000 | 5000 |
| B | 30000 | 7000 |
| C | 40000 | 10000 |

The company can add the salary and bonus to calculate total compensation.

Business use:
- Calculate employee compensation.
- Calculate combined sales.
- Analyse total customer spending.

AI/ML use:
A combined feature can provide a more meaningful representation of total income or spending.


### Step 3 – Implement the Concept

Create a Total_Salary feature by adding Basic_Salary and Bonus.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C"],
    "Basic_Salary": [25000, 30000, 40000],
    "Bonus": [5000, 7000, 10000]
})

df["Total_Salary"] = df["Basic_Salary"] + df["Bonus"]

print(df)
```

### Output Explanation

The Total_Salary column contains the sum of Basic_Salary and Bonus.

Examples:

Employee A: 25000 + 5000 = 30000

Employee B: 30000 + 7000 = 37000

Employee C: 40000 + 10000 = 50000

The new feature represents the employee's total salary including bonus.

## Topic 4 – Feature − Feature

### Step 1 – Understand the Concept

Feature − Feature means subtracting one numerical feature from another to create a new feature.

Formula:

New Feature = Feature 1 − Feature 2

Subtraction helps identify differences between two values.

Examples:
- Selling Price − Cost Price
- Current Balance − Previous Balance
- Actual Sales − Target Sales
- Current Stock − Previous Stock

The resulting feature represents the difference between two values.



### Step 2 – Demonstrate the Concept

Real-world example:

A retail company wants to calculate the profit earned from product sales.

| Product | Selling Price | Cost Price |
|---|---:|---:|
| Laptop | 50000 | 40000 |
| Mobile | 20000 | 15000 |
| Tablet | 15000 | 12000 |

The company subtracts Cost Price from Selling Price.

Business use:
- Calculate product profit.
- Identify sales performance.
- Compare actual and expected values.

AI/ML use:
Difference features can help models identify changes, gaps, and performance patterns.


### Step 3 – Implement the Concept

Create a Profit feature by subtracting Cost_Price from Selling_Price.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Selling_Price": [50000, 20000, 15000],
    "Cost_Price": [40000, 15000, 12000]
})

df["Profit"] = df["Selling_Price"] - df["Cost_Price"]

print(df)
```

### Output Explanation

The Profit column represents the difference between Selling_Price and Cost_Price.

Examples:

Laptop: 50000 - 40000 = 10000

Mobile: 20000 - 15000 = 5000

Tablet: 15000 - 12000 = 3000

A positive result represents profit in this example.

## Topic 5 – Feature / Feature

### Step 1 – Understand the Concept

Feature / Feature means dividing one numerical feature by another numerical feature.

Formula:

New Feature = Feature 1 / Feature 2

Division is useful for creating ratios, averages, and relative measurements.

Examples:
- Total Spending / Number of Orders
- Experience / Age
- Total Distance / Total Time
- Total Revenue / Number of Customers

These features help understand the relationship between two numerical values.

Important:
We must handle zero values in the denominator to avoid division errors.



### Step 2 – Demonstrate the Concept

Real-world example:

An online shopping company wants to calculate the average spending per order.

| Customer | Total Spending | Number of Orders |
|---|---:|---:|
| A | 10000 | 5 |
| B | 18000 | 6 |
| C | 12000 | 4 |

Formula:

Average Order Value = Total Spending / Number of Orders

Business use:
- Understand customer spending.
- Compare customer purchasing behaviour.
- Identify high-value customers.

AI/ML use:
Ratio features can help models compare customers with different spending amounts and order counts.


### Step 3 – Implement the Concept

Create an Average_Order_Value feature by dividing Total_Spending by Number_of_Orders.

We will also handle zero values safely.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Customer": ["A", "B", "C"],
    "Total_Spending": [10000, 18000, 12000],
    "Number_of_Orders": [5, 6, 4]
})

df["Average_Order_Value"] = (
    df["Total_Spending"] /
    df["Number_of_Orders"].replace(0, np.nan)
)

print(df)
```

### Output Explanation

The Average_Order_Value column represents the average amount spent per order.

Examples:

Customer A: 10000 / 5 = 2000

Customer B: 18000 / 6 = 3000

Customer C: 12000 / 4 = 3000

If the number of orders is zero, the result becomes NaN instead of causing a division-by-zero error.

## Topic 6 – Polynomial Features

### Step 1 – Understand the Concept

Polynomial Features are new features created by raising existing numerical features to different powers or by multiplying features together.

Examples:

- Age²
- Income²
- Age × Income
- Experience²

Formula:

X² = X × X

X³ = X × X × X

Polynomial features help Machine Learning models learn nonlinear relationships.

They are commonly used with Linear Regression when the relationship between input features and the target is not a straight line.



### Step 2 – Demonstrate the Concept

Real-world example:

A company wants to understand how employee experience affects salary.

Salary may not increase at the same rate for every additional year of experience.

For example:
- Employees with 1 year of experience may receive a small salary increase.
- Employees with 10 years of experience may receive a larger salary increase.
- Salary growth may slow after a certain experience level.

A polynomial feature such as Experience² can help represent this curved relationship.

Business use:
- Analyse salary growth.
- Understand nonlinear sales patterns.
- Model changing customer behaviour.

AI/ML use:
Polynomial features allow models such as Linear Regression to learn more complex patterns.



### Step 3 – Implement the Concept

We will create:
- Experience_Squared
- Experience_Cubed

We will also use PolynomialFeatures from Scikit-learn to generate polynomial combinations.

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5]
})

df["Experience_Squared"] = df["Experience"] ** 2
df["Experience_Cubed"] = df["Experience"] ** 3

poly = PolynomialFeatures(degree=2, include_bias=False)

poly_values = poly.fit_transform(df[["Experience"]])

poly_df = pd.DataFrame(
    poly_values,
    columns=poly.get_feature_names_out(["Experience"])
)

print("Original and manually created features:")
print(df)

print("\nPolynomial features:")
print(poly_df)
```

### Output Explanation

Experience_Squared contains the square of Experience.

Experience_Cubed contains the cube of Experience.

For example:

Experience = 3

3² = 9

3³ = 27

PolynomialFeatures with degree=2 creates:
- Experience
- Experience²

Polynomial features help represent nonlinear patterns.

However, creating too many polynomial features can increase model complexity and cause overfitting.

## Topic 7 – Pairwise Interactions

### Step 1 – Understand the Concept

Pairwise Interactions are features created by combining pairs of existing features.

A pair means two features are combined at a time.

For example, if a dataset contains:

- Age
- Income
- Experience

Possible pairwise interactions include:

- Age × Income
- Age × Experience
- Income × Experience

These combinations help a model understand how two variables work together.

Pairwise interaction features are commonly generated using PolynomialFeatures with interaction_only=True.



### Step 2 – Demonstrate the Concept

Real-world example:

A bank wants to predict whether a customer may be eligible for a particular loan product.

The bank has information about:
- Customer Age
- Annual Income
- Work Experience

Age and Income together may provide additional information about the customer's financial profile.

Business use:
- Customer segmentation.
- Loan analysis.
- Risk assessment.

AI/ML use:
Pairwise interaction features help a model identify relationships between two input variables that may influence the prediction.



### Step 3 – Implement the Concept

We will create pairwise interaction features using Scikit-learn.

The interaction_only=True option creates pairwise products without creating squared features.

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

df = pd.DataFrame({
    "Age": [22, 30, 40],
    "Income": [30000, 50000, 80000],
    "Experience": [1, 5, 15]
})

interaction = PolynomialFeatures(
    degree=2,
    interaction_only=True,
    include_bias=False
)

interaction_values = interaction.fit_transform(df)

interaction_df = pd.DataFrame(
    interaction_values,
    columns=interaction.get_feature_names_out(df.columns)
)

print(interaction_df)
```

### Output Explanation

The output contains the original features and pairwise interaction features.

Examples:

Age × Income

Age × Experience

Income × Experience

The model can use these additional columns to learn combined relationships.

The interaction_only=True option avoids creating squared terms such as Age² and Income².

## Topic 8 – Domain-Based Interactions

### Step 1 – Understand the Concept

Domain-Based Interactions are new features created using business knowledge or knowledge about a particular industry.

Instead of combining features randomly, we select combinations that have a meaningful business purpose.

Examples:

Banking:
- Loan Amount / Annual Income
- Monthly Debt / Monthly Income

E-commerce:
- Quantity × Price
- Total Spending / Number of Orders

Healthcare:
- Weight / Height²
- Blood Pressure Difference

Employee Analytics:
- Experience / Age
- Salary / Experience

Domain knowledge helps us create features that represent meaningful real-world relationships.


### Step 2 – Demonstrate the Concept

Real-world example:

An e-commerce company wants to understand customer purchasing behaviour.

It has the following information:

- Total Spending
- Number of Orders
- Total Quantity Purchased
- Average Product Price

The company creates:

1. Average Order Value = Total Spending / Number of Orders

2. Average Quantity per Order = Total Quantity Purchased / Number of Orders

3. Total Purchase Value = Quantity × Average Product Price

Business use:
- Understand customer value.
- Identify frequent buyers.
- Analyse product purchasing patterns.

AI/ML use:
Domain-based interaction features can improve predictions because they represent meaningful relationships from the business domain.


### Step 3 – Implement the Concept

We will create domain-based interaction features for an e-commerce dataset.

The new features will represent:
- Average Order Value
- Average Quantity per Order
- Total Purchase Value

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Customer": ["A", "B", "C"],
    "Total_Spending": [10000, 18000, 12000],
    "Number_of_Orders": [5, 6, 4],
    "Total_Quantity": [20, 30, 16],
    "Average_Product_Price": [500, 600, 750]
})

df["Average_Order_Value"] = (
    df["Total_Spending"] /
    df["Number_of_Orders"].replace(0, np.nan)
)

df["Average_Quantity_Per_Order"] = (
    df["Total_Quantity"] /
    df["Number_of_Orders"].replace(0, np.nan)
)

df["Total_Purchase_Value"] = (
    df["Total_Quantity"] *
    df["Average_Product_Price"]
)

print(df)
```

### Output Explanation

The output contains three new domain-based interaction features.

Average_Order_Value:
Represents the average amount spent per order.

Average_Quantity_Per_Order:
Represents the average number of products purchased per order.

Total_Purchase_Value:
Represents the estimated purchase value based on quantity and average product price.

These features provide meaningful information about customer purchasing behaviour.

## When Are Interaction Features Useful?

Interaction features are useful when:

1. Two or more features together provide meaningful information.
2. The relationship between features influences the target variable.
3. Business knowledge suggests that a combination of features is important.
4. A Machine Learning model needs additional information to identify complex patterns.
5. We want to create meaningful ratios, differences, totals, or product combinations.

Examples:

- Income × Age can help analyse customer purchasing behaviour.
- Quantity × Price can calculate purchase value.
- Experience / Age can represent experience relative to age.
- Total Spending / Number of Orders can calculate average order value.

## When Can Interaction Features Create Unnecessary Complexity?

Interaction features can create unnecessary complexity when:

1. Too many combinations are generated without a clear purpose.
2. New features contain duplicate or redundant information.
3. The dataset is small but the number of features becomes very large.
4. Features introduce noise instead of useful information.
5. The model starts overfitting the training data.
6. Features use information that would not be available at prediction time, causing data leakage.

## Important Considerations

- Create features based on business knowledge.
- Check whether the new features are meaningful.
- Avoid unnecessary combinations.
- Handle zero values when creating ratios.
- Check for missing and infinite values.
- Evaluate model performance before and after adding interaction features.
- Use feature selection or regularization when necessary.

Conclusion:

Interaction features help Machine Learning models understand relationships between multiple variables.

However, creating too many unnecessary interaction features can increase model complexity, training time, and overfitting.
