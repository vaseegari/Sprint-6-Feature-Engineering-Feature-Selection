# Topic 1 – What is a Feature?

## Step 1 – Understand the Concept

### Definition

A feature is an individual piece of information or a column in a dataset that describes a person, product, transaction, or object.

Features are used as input information by Machine Learning models to identify patterns and make predictions.

### Example

Consider a customer dataset:

- Age
- Monthly Income
- Monthly Expenses
- Customer Tenure

These columns are called features because they provide information about customers.

### Why are Features Important?

1. They provide input information to Machine Learning models.
2. They help models identify patterns.
3. They provide useful information for predictions.
4. They help models understand the data.

### Conclusion

Features are important parts of a dataset because they provide the information required by Machine Learning models to learn and make predictions.

## Topic 2 – Feature vs Variable

## Step 1 – Understand the Concept

### What is a Variable?

A variable is a general term used to represent any characteristic or value that can change.

### What is a Feature?

A feature is a variable specifically used as an input to a Machine Learning model.

### Difference Between Feature and Variable

| Variable | Feature |
|---|---|
| General data characteristic | Input used by an ML model |
| Can represent different types of information | Selected for model learning |
| May include input and target information | Represents model input |
| Used in programming and statistics | Commonly used in Machine Learning |

### Example

Consider a customer dataset:

- Age
- Monthly Income
- Monthly Expenses
- Churn Status

All these columns are variables in the dataset.

If we build a customer churn prediction model:

- Age, Monthly Income, Monthly Expenses are features.
- Churn Status is the target.

### Important Point

Every feature is a variable, but every variable is not necessarily a feature in a particular Machine Learning problem.




## Step 2 – Demonstrate the Concept

### 1. Simple Example

In a student dataset:

- Student Age
- Study Hours
- Exam Marks

All are variables.

If we predict exam marks, Student Age and Study Hours can be selected as features.

### 2. Real-World Example

In an e-commerce dataset:

- Product Price
- Quantity
- Total Amount

These are variables.

If we predict Total Amount, Product Price and Quantity are the input features.

### 3. Business Example

A bank stores customer income, expenses, loan amount, and loan repayment status.

For a loan default prediction model, income, expenses, and loan amount can be features. Repayment status is the target.

### 4. AI/ML Use Case

Feature selection depends on the prediction problem.

The same variable can be a feature in one ML problem and a target in another problem.



## Step 3 – Implement the Concept

```python
data = {
    'Age': [23, 30, 35, 40, 28],
    'Monthly_Income': [20000, 30000, 40000, 50000, 25000],
    'Monthly_Expenses': [12000, 18000, 25000, 30000, 15000],
    'Churn_Status': [0, 0, 1, 1, 0]
}

df_variable = pd.DataFrame(data)

df_variable
```

4. Churn Status is used as the target in this example.

### Observation

Variables represent all available information, while features represent the selected input information for a specific ML problem.

## Topic 4 – What is Feature Engineering?

## Step 1 – Understand the Concept

### Definition

Feature Engineering is the process of creating new features or modifying existing features from available data to make the information more useful for Machine Learning models.

It helps transform raw data into meaningful information.

### Example

Suppose we have:

- Monthly Income = 30,000
- Monthly Expenses = 18,000

We can create a new feature:

Savings = Monthly Income - Monthly Expenses

Savings = 12,000

The new feature provides additional information about the customer's financial behaviour.

### Main Activities

1. Feature Creation
2. Feature Transformation
3. Feature Selection
4. Feature Extraction

### Why is Feature Engineering Important?

- Creates meaningful information from existing data.
- Helps models identify useful patterns.
- Can improve model performance.
- Converts business knowledge into useful ML inputs.

### Conclusion

Feature Engineering prepares useful and meaningful information for Machine Learning models.



## Step 2 – Demonstrate the Concept

### 1. Simple Example

Income = 30,000

Expenses = 18,000

Savings = 12,000

A new Savings feature is created using existing columns.

### 2. Real-World Example

In an e-commerce application:

Product Price = 500

Quantity = 3

Total Amount = 500 × 3 = 1,500

### 3. Business Example

In banking, monthly income and monthly debt can be used to calculate a Debt-to-Income Ratio.

### 4. AI/ML Use Case

In customer churn prediction, customer tenure and average monthly spending may provide useful information to identify customer behaviour.

### Why is this useful?

The new features can reveal information that was not directly available in the original columns.



## Step 3 – Implement the Concept

```python
data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Monthly_Income': [20000, 30000, 40000, 50000, 25000],
    'Monthly_Expenses': [12000, 18000, 25000, 30000, 15000]
}

df_engineering = pd.DataFrame(data)

df_engineering
```

```python
df_engineering['Savings'] = (
    df_engineering['Monthly_Income']
    - df_engineering['Monthly_Expenses']
)

df_engineering
```

```python
df_engineering['Savings_Ratio'] = (
    df_engineering['Savings']
    / df_engineering['Monthly_Income']
)

df_engineering
```

### Code Explanation

1. Created a sample customer dataset.
2. Used Monthly Income and Monthly Expenses to calculate Savings.
3. Created Savings_Ratio by dividing Savings by Monthly Income.
4. Displayed the updated dataset.

### Feature Justification

Feature Name: Savings

Source Columns: Monthly_Income, Monthly_Expenses

Logic: Income - Expenses

Reason: To understand the remaining money after expenses.

Business Meaning: Customer saving capacity.

ML Relevance: May help understand financial behaviour.

Leakage Risk: Low if both source columns are available at prediction time.

Final Decision: Retain for further analysis.

Feature Name: Savings_Ratio

Source Columns: Savings, Monthly_Income

Logic: Savings / Monthly Income

Reason: To understand the proportion of income saved.

Business Meaning: Customer saving percentage.

ML Relevance: May provide additional financial information.

Leakage Risk: Low if source values are available at prediction time.

Final Decision: Retain for further analysis.

## Topic 7 – Good Features vs Bad Features

## Step 1 – Understand the Concept

### What are Good Features?

Good features provide useful and meaningful information to a Machine Learning model.

### Characteristics of Good Features

1. Relevant to the target.
2. Provide meaningful information.
3. Available at prediction time.
4. Have reasonable data quality.
5. Do not introduce data leakage.

### What are Bad Features?

Bad features may contain irrelevant, redundant, misleading, or leaked information.

### Characteristics of Bad Features

1. Irrelevant to the prediction problem.
2. Contain duplicate information.
3. Have no useful predictive information.
4. Contain target leakage.
5. Are unavailable at prediction time.

### Example – Customer Churn Prediction

Good Features:
- Customer Tenure

- Contract Type
- Payment Method

Potentially Bad Features:
- Random Customer Reference Number
- Duplicate Columns

A feature's usefulness depends on the business problem, data quality, and prediction time.

- Helps identify missing information.

- Prevents data leakage.

### Conclusion

A proper workflow helps prepare meaningful and reliable features for Machine Learning models.

## Step 3 – Implement the Concept

```python
data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Monthly_Income': [20000, 30000, 40000, 50000, 25000],
    'Monthly_Expenses': [12000, 18000, 25000, 30000, 15000]
}

df_workflow = pd.DataFrame(data)

df_workflow
```

## Step 3 – Implement the Concept

```python
data = {
    'Product_Price': [500, 1000, 750, 1200, 300],
    'Quantity': [2, 1, 4, 2, 5],
    'Previous_Amount': [900, 800, 2500, 2000, 1000]
}

df_creation = pd.DataFrame(data)

df_creation
```

```python
df_creation['Total_Amount'] = (
    df_creation['Product_Price']
    * df_creation['Quantity']
)

df_creation['Amount_Difference'] = (
    df_creation['Total_Amount']
    - df_creation['Previous_Amount']
)

df_creation['Price_Per_Quantity'] = (
    df_creation['Total_Amount']
    / df_creation['Quantity']
)

df_creation
```

### Code Explanation

1. Created an e-commerce dataset.
2. Multiplied product price by quantity.
3. Created Total_Amount.
4. Subtracted Previous_Amount from Total_Amount.
5. Created Amount_Difference.
6. Divided Total_Amount by Quantity.
7. Created Price_Per_Quantity.

### Feature Justification

Total_Amount:
- Source: Product_Price, Quantity
- Logic: Price × Quantity
- Business Meaning: Total purchase value
- ML Relevance: Customer spending information
- Leakage Risk: Low if both values are available at prediction time
- Decision: Retain for analysis

Amount_Difference:
- Source: Total_Amount, Previous_Amount
- Logic: Total Amount - Previous Amount
- Business Meaning: Difference between current and previous transaction values
- ML Relevance: May show changes in spending
- Leakage Risk: Depends on whether both amounts are available at prediction time
- Decision: Needs further analysis

Price_Per_Quantity:
- Source: Total_Amount, Quantity
- Logic: Total Amount / Quantity
- Business Meaning: Average price per unit
- ML Relevance: Provides unit price information
- Leakage Risk: Low if source values are available
- Decision: Retain for analysis

## Topic 10 – Feature Transformation


## Step 1 – Understand the Concept

Feature Transformation is the process of changing the representation or scale of existing features to make them more suitable for Machine Learning algorithms.

It changes how information is represented without necessarily creating entirely new information.

### Common Techniques

1. Scaling
2. Normalization
3. Standardization
4. Log Transformation
5. Encoding

### Example

Original Income Values:

1000, 2000, 3000, 4000, 5000

After Min-Max Normalization:

0.00, 0.25, 0.50, 0.75, 1.00

### Why is Feature Transformation Useful?

- Handles different numerical scales.
- Helps reduce skewness.
- Converts categorical data into numerical values.
- Makes data suitable for specific algorithms.

### Conclusion

Feature Transformation changes the representation of existing features to help Machine Learning models process information effectively.



## Step 2 – Demonstrate the Concept

### 1. Simple Example

Suppose two features have different scales:

Age: 20 to 60

Income: 10,000 to 100,000

We can use scaling to bring numerical values into a common range.

### 2. Real-World Example

In banking, customer income values may be much larger than customer age values.

Scaling can help algorithms that are sensitive to feature magnitudes.

### 3. Business Example

An e-commerce company may have product prices ranging from 100 to 100,000.

Normalization can represent these values within a common range.

### 4. AI/ML Use Case

Scaling is useful for algorithms such as:

- K-Nearest Neighbors
- Support Vector Machines
- Logistic Regression

### Why is it useful?

It helps prevent features with larger numerical scales from dominating distance-based or optimization-based learning methods.




## Step 3 – Implement the Concept

```python
data = {
    'Age': [20, 25, 30, 40, 50],
    'Monthly_Income': [10000, 25000, 40000, 70000, 100000]
}

df_transform = pd.DataFrame(data)

df_transform
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_values = scaler.fit_transform(df_transform)

df_scaled = pd.DataFrame(
    scaled_values,
    columns=df_transform.columns
)

df_scaled
```

```python
print("Original Dataset:")
display(df_transform)

print("Transformed Dataset:")
display(df_scaled)
```

### Code Explanation

1. Imported MinMaxScaler from Scikit-learn.
2. Created a sample dataset containing Age and Monthly Income.
3. Created a scaler object.
4. Used fit_transform() to learn the minimum and maximum values and transform the data.
5. Converted the transformed values into a DataFrame.
6. Compared the original and transformed datasets.

### Observation

## Topic 13 – Feature Leakage


## Step 1 – Understand the Concept

### What is Feature Leakage?

Feature Leakage occurs when a Machine Learning model receives information during training that would not actually be available when making real-world predictions.

This can make model performance appear better during training or testing than it will be in real-world usage.

### Types of Feature Leakage

1. Target Leakage
2. Train-Test Leakage
3. Temporal Leakage
4. Post-Outcome Features
5. Leakage Through Aggregations
6. Leakage Through Target Encoding

### Example – Customer Churn

Suppose we want to predict whether a customer will leave a company.

Available Columns:

- Customer Tenure
- Monthly Charges
- Contract Type
- Account Closure Date

If Account Closure Date is recorded only after the customer has left, using it as an input feature introduces leakage.

The model receives information about an event that it is supposed to predict.

### Why Should We Prevent Leakage?

1. Prevents misleading model performance.
2. Improves real-world reliability.
3. Ensures fair model evaluation.
4. Helps build trustworthy Machine Learning systems.

### Conclusion

Always check whether a feature would genuinely be available at the time of prediction.



## Step 2 – Demonstrate the Concept

### 1. Simple Example

Suppose we want to predict exam results.

Features:
- Study Hours
- Attendance

Target:
- Exam Result

If we use the final exam result itself as an input feature, it creates target leakage.

### 2. Real-World Example

In banking, if we predict loan default, we should not use information recorded after the customer has defaulted.

### 3. Business Example

A telecom company predicts customer churn.

If account closure information is recorded after a customer leaves, using it to predict churn introduces leakage.

### 4. AI/ML Use Case

A churn prediction model should only use information available before the prediction date.

### Why is it useful?

Preventing leakage helps the model learn genuine patterns instead of using future information.



## Step 3 – Implement the Concept

```python
data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Monthly_Charges': [500, 700, 900, 600, 800],
    'Customer_Tenure': [6, 12, 24, 36, 10],
    'Churn_Status': [0, 0, 1, 1, 0],
    'Account_Closed': [0, 0, 1, 1, 0]
}

df_leakage = pd.DataFrame(data)

df_leakage
```

```python
target = 'Churn_Status'

potential_leakage = 'Account_Closed'

print("Target Variable:", target)

print("Potential Leakage Feature:", potential_leakage)
```

```python
X_correct = df_leakage[
    ['Monthly_Charges', 'Customer_Tenure']
]

y = df_leakage['Churn_Status']

print("Correct Input Features:")
display(X_correct)

print("Target:")
display(y)
```

### Code Explanation

1. Created a sample customer churn dataset.
2. Included monthly charges and customer tenure.
3. Created Churn_Status as the target.
4. Added Account_Closed as a potential leakage column.
5. Identified the potential leakage feature.
6. Selected only the customer information columns as model inputs.

### Feature Leakage Analysis

Feature Name: Account_Closed

Source Column: Account_Closed

Reason for Removal: It may contain information recorded after the customer has already left.

Business Meaning: Indicates whether the customer account is closed.

ML Relevance: It may reveal the outcome instead of helping predict it.

Leakage Risk: High if it is recorded after churn.

Final Decision: Remove from the prediction features.

### Important Point

This is a simplified example. In a real dataset, feature availability must be checked using timestamps and business rules.

```python
data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Monthly_Income': [20000, 30000, 40000, 50000, 25000],
    'Monthly_Expenses': [12000, 18000, 25000, 30000, 15000],
    'Customer_Tenure': [6, 12, 24, 36, 10]
}

df_final = pd.DataFrame(data)

# Feature Engineering
df_final['Savings'] = (
    df_final['Monthly_Income']
    - df_final['Monthly_Expenses']
)

df_final['Expense_Ratio'] = (
    df_final['Monthly_Expenses']
    / df_final['Monthly_Income']
)

# Select final features
final_features = df_final[
    ['Customer_Tenure', 'Savings', 'Expense_Ratio']
]

display(final_features)
```
