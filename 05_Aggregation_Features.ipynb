## Topic 1 – Count


### Step 1: Understand the Concept

Count Aggregation calculates the number of records or non-missing values in a dataset or group.

In Pandas:
- count() counts non-missing values.
- size() counts all rows, including rows with missing values.

Count aggregation helps identify how many times an activity occurs.

Examples:
- Number of customer purchases
- Number of product orders
- Number of transactions


### Step 2: Demonstrate the Concept

### Real-World Example

An e-commerce company wants to calculate how many purchases each customer has made.

Example:

| Customer_ID | Order_ID |
|---|---|
| 101 | 1 |
| 101 | 2 |
| 102 | 3 |
| 102 | 4 |
| 102 | 5 |
| 103 | 6 |

Expected Aggregated Result:

| Customer_ID | Purchase_Count |
|---|---:|
| 101 | 2 |
| 102 | 3 |
| 103 | 1 |

Business Use:
- Identify frequent customers.
- Analyze customer activity.
- Create customer loyalty programs.

AI/ML Use:
Purchase count can help machine learning models identify customer behavior and predict repeat purchases.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 102, 102, 102, 103],
    "Order_ID": [1, 2, 3, 4, 5, 6],
    "Purchase_Amount": [1000, 2000, 1500, 2500, 3000, 4000]
})

# Count purchases for each customer
purchase_count = (
    df.groupby("Customer_ID")
    .size()
    .reset_index(name="Purchase_Count")
)

print(purchase_count)
```

### Output Explanation

The groupby() function groups transactions based on Customer_ID.

The size() function counts the total number of rows in each customer group.

Expected Result:

- Customer 101: 2 purchases
- Customer 102: 3 purchases
- Customer 103: 1 purchase

This aggregation feature helps businesses understand how frequently customers make purchases.

## Topic 2 – Sum


### Step 1: Understand the Concept

Sum Aggregation calculates the total value of a numerical column.

Formula:

Total = Value 1 + Value 2 + ... + Value N

In Pandas, the sum() function calculates the total.

Examples:
- Total customer spending
- Total sales revenue
- Total product quantity sold
- Total monthly expenses


### Step 2: Demonstrate the Concept

### Real-World Example

An online shopping company wants to calculate the total amount spent by each customer.

Example:

Customer 101:
- Purchase 1: ₹1000
- Purchase 2: ₹2000
- Purchase 3: ₹1500

Total Spending = ₹4500

Business Use:
- Calculate customer spending.
- Measure total sales.
- Analyze company revenue.

AI/ML Use:
Total customer spending can help identify high-value customers and predict future purchasing behavior.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 101, 102, 102, 103],
    "Purchase_Amount": [1000, 2000, 1500, 3000, 2500, 4000]
})

# Calculate total spending for each customer
total_spending = (
    df.groupby("Customer_ID")["Purchase_Amount"]
    .sum()
    .reset_index(name="Total_Spending")
)

print(total_spending)
```

### Output Explanation

The groupby() function groups purchase records by Customer_ID.

The sum() function calculates the total purchase amount for each customer.

Expected Result:

- Customer 101: ₹4500
- Customer 102: ₹5500
- Customer 103: ₹4000

This feature helps businesses understand the total amount spent by each customer.

## Topic 3 – Mean


### Step 1: Understand the Concept

Mean Aggregation calculates the average value of a numerical column.

Formula:

Mean = Sum of Values / Number of Values

Example:

Purchase Amounts: 1000, 2000, 3000

Mean = (1000 + 2000 + 3000) / 3

Mean = 2000

In Pandas, the mean() function calculates the average.

Mean aggregation helps understand the typical value in a dataset.



### Step 2: Demonstrate the Concept


### Real-World Example

An e-commerce company wants to calculate the average order value of each customer.

Example:

Customer 101:
- Order 1: ₹1000
- Order 2: ₹2000
- Order 3: ₹3000

Average Order Value = ₹2000

Business Use:
- Average order value analysis.
- Average customer spending.
- Sales performance analysis.

AI/ML Use:
Average spending can help identify customer spending patterns and predict future purchases.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 101, 102, 102, 103],
    "Purchase_Amount": [1000, 2000, 3000, 1500, 2500, 4000]
})

# Calculate average purchase amount
average_purchase = (
    df.groupby("Customer_ID")["Purchase_Amount"]
    .mean()
    .reset_index(name="Average_Order_Value")
)

print(average_purchase)
```

### Output Explanation

The mean() function calculates the average purchase amount for each customer.

Expected Result:

- Customer 101: ₹2000
- Customer 102: ₹2000
- Customer 103: ₹4000

The average order value helps businesses understand how much customers spend on an average purchase.

## Topic 4 – Median

### Step 1: Understand the Concept

Median is the middle value of a dataset after arranging the values in ascending order.

If the number of values is odd, the median is the middle value.

If the number of values is even, the median is the average of the two middle values.

Example 1:

Values: 100, 200, 300

Median = 200

Example 2:

Values: 100, 200, 300, 400

Median = (200 + 300) / 2

Median = 250

In Pandas, the median() function calculates the median value.



### Step 2: Demonstrate the Concept

### Real-World Example

An online shopping company wants to find the typical purchase amount of its customers.

Suppose the purchase amounts are:

₹100, ₹200, ₹300, ₹400, ₹10000

The mean is affected by the unusually large purchase of ₹10000.

The median is ₹300 and gives a better representation of the middle purchase amount in this example.

Business Use:
- Typical customer spending analysis.
- Product price analysis.
- Income and transaction analysis.

AI/ML Use:
Median aggregation can help create customer features that are less affected by unusually large or small transactions.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 101, 101, 101, 102, 102, 102],
    "Purchase_Amount": [100, 200, 300, 400, 10000, 500, 700, 900]
})

# Calculate median purchase amount
median_purchase = (
    df.groupby("Customer_ID")["Purchase_Amount"]
    .median()
    .reset_index(name="Median_Purchase")
)

print(median_purchase)
```

### Output Explanation

The median() function calculates the middle purchase amount for each customer.

Expected Result:

- Customer 101: ₹300
- Customer 102: ₹700

Median is less affected by extreme values than the mean.

Therefore, it can be useful when customer transaction amounts contain outliers.

## Topic 5 – Minimum


### Step 1: Understand the Concept

Minimum Aggregation identifies the smallest value in a numerical column.

In Pandas, the min() function returns the minimum value.

Example:

Purchase Amounts: 1000, 2000, 3000, 500

Minimum = 500

It helps identify the lowest transaction amount, lowest price, or lowest recorded value.


### Step 2: Demonstrate the Concept


### Real-World Example

An e-commerce company wants to find the lowest purchase amount made by each customer.

Business Use:
- Lowest customer transaction.
- Lowest product price.
- Minimum monthly expenses.

AI/ML Use:
Minimum transaction amount can help understand customer spending ranges and identify unusual transaction patterns.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 101, 102, 102, 103],
    "Purchase_Amount": [1000, 500, 2000, 1500, 3000, 800]
})

# Calculate minimum purchase amount
minimum_purchase = (
    df.groupby("Customer_ID")["Purchase_Amount"]
    .min()
    .reset_index(name="Minimum_Purchase")
)

print(minimum_purchase)
```

The min() function identifies the smallest purchase amount for each customer.

## Topic 7 – Standard Deviation


### Step 1: Understand the Concept

Standard Deviation measures how much the values in a dataset are spread out from the mean.

A low standard deviation means the values are close to the mean.

A high standard deviation means the values are more spread out.

In Pandas, the std() function calculates the standard deviation.

By default, Pandas uses sample standard deviation (ddof=1).


### Step 2: Demonstrate the Concept


### Real-World Example

An e-commerce company compares the purchase behavior of two customers.

Customer A:
1000, 1100, 1050

Customer B:
100, 2000, 5000

Customer A has more consistent purchase amounts.

Customer B has more variation in purchase amounts.

Business Use:
- Understand customer spending consistency.
- Analyze sales variation.
- Compare product price variations.

AI/ML Use:
Standard deviation can help machine learning models understand customer behavior variation and identify unusual patterns.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [
        101, 101, 101,
        102, 102, 102
    ],
    "Purchase_Amount": [
        1000, 1100, 1050,
        100, 2000, 5000
    ]
})

# Calculate standard deviation for each customer
spending_variation = (
    df.groupby("Customer_ID")["Purchase_Amount"]
    .std()
    .reset_index(name="Purchase_Standard_Deviation")
)

print(spending_variation)
```

### Output Explanation

The std() function calculates the spread of purchase amounts around the mean.

Expected observation:

- Customer 101 has a lower standard deviation.
- Customer 102 has a higher standard deviation.

This means Customer 101 has more consistent spending, while Customer 102 has more variation.

Standard deviation helps businesses understand how consistent or variable customer purchases are.

## Topic 8 – Group-Based Aggregation


### Step 1: Understand the Concept

Group-Based Aggregation combines records into groups using one or more columns and calculates summary values for each group.

The Pandas groupby() function is used for this operation.

Common aggregation functions:

- count()
- sum()
- mean()
- median()
- min()
- max()
- std()

Group-based aggregation helps compare different categories, regions, departments, or customer groups.


### Step 2: Demonstrate the Concept


### Real-World Example

A retail company sells products in different regions.

The company wants to calculate total sales and average sales for each region.

Business Use:
- Region-wise sales analysis.
- Department performance.
- Category-wise revenue analysis.

AI/ML Use:
Group-based features can help machine learning models identify differences between regions and customer segments.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Region": [
        "Chennai", "Chennai",
        "Mumbai", "Mumbai",
        "Delhi", "Delhi"
    ],
    "Sales": [10000, 15000, 20000, 25000, 12000, 18000]
})

# Calculate multiple aggregations for each region
region_summary = (
    df.groupby("Region")["Sales"]
    .agg(["count", "sum", "mean", "min", "max"])
    .reset_index()
)

print(region_summary)
```

### Output Explanation

The groupby("Region") function creates separate groups for each region.

The agg() function applies multiple aggregation operations.

The resulting columns represent:

- count: Number of sales records.
- sum: Total sales.
- mean: Average sales.
- min: Lowest sales value.
- max: Highest sales value.

This approach provides multiple business insights from a single grouped dataset.

## Topic 9 – Customer-Level Aggregation


### Step 1: Understand the Concept

Customer-Level Aggregation summarizes transaction data for each customer.

It converts multiple transaction records into customer-level features.

Common customer-level features:

- Total customer spending.
- Average order value.
- Number of purchases.
- Maximum transaction amount.
- Minimum transaction amount.

These features help understand individual customer behavior.


### Step 2: Demonstrate the Concept

### Real-World Example

An online shopping company stores thousands of customer transactions.

The company wants to create a summary for each customer.

Example:

Customer 101:
- Total Spending: ₹4500
- Average Order Value: ₹1500
- Purchase Count: 3
- Maximum Purchase: ₹2000

Business Use:
- Customer segmentation.
- Loyalty program planning.
- Identifying high-value customers.

AI/ML Use:
Customer-level aggregation features can be used for customer churn prediction, customer segmentation, and purchase prediction.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 101, 102, 102, 103],
    "Transaction_ID": [1, 2, 3, 4, 5, 6],
    "Purchase_Amount": [1000, 2000, 1500, 3000, 2500, 4000]
})

# Create customer-level aggregated features
customer_features = (
    df.groupby("Customer_ID")
    .agg(
        Total_Spending=("Purchase_Amount", "sum"),
        Average_Order_Value=("Purchase_Amount", "mean"),
        Purchase_Count=("Transaction_ID", "count"),
        Maximum_Transaction=("Purchase_Amount", "max"),
        Minimum_Transaction=("Purchase_Amount", "min")
    )
    .reset_index()
)

print(customer_features)
```

### Output Explanation

The groupby() function groups transactions by Customer_ID.

The agg() function calculates multiple customer-level features.

The resulting features include:

- Total_Spending: Total amount spent by the customer.
- Average_Order_Value: Average transaction amount.
- Purchase_Count: Number of transactions.
- Maximum_Transaction: Highest transaction amount.
- Minimum_Transaction: Lowest transaction amount.

The final dataset contains one row per customer.

These features can be used as inputs for machine learning models.

## Topic 10 – Product-Level Aggregation

### Step 1: Understand the Concept

Product-Level Aggregation summarizes sales and transaction information for each product.

It combines multiple product records into useful product-level features.

Common product-level features:

- Total quantity sold.
- Total product revenue.
- Average product rating.
- Number of orders.
- Maximum selling price.

These features help businesses understand product performance.


### Step 2: Demonstrate the Concept


### Real-World Example

An online shopping company wants to identify its best-performing products.

The company calculates:

- Total quantity sold.
- Total revenue.
- Average product rating.
- Number of orders.

Business Use:
- Identify popular products.
- Manage product inventory.
- Compare product performance.

AI/ML Use:
Product-level features can help recommend products and predict product demand.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Product_ID": ["P101", "P101", "P102", "P102", "P103"],
    "Quantity": [2, 3, 1, 4, 2],
    "Unit_Price": [500, 500, 1000, 1000, 750],
    "Rating": [4.5, 5.0, 4.0, 4.5, 3.5]
})

# Calculate revenue for each transaction
df["Revenue"] = df["Quantity"] * df["Unit_Price"]

# Calculate product-level features
product_features = (
    df.groupby("Product_ID")
    .agg(
        Total_Quantity_Sold=("Quantity", "sum"),
        Total_Revenue=("Revenue", "sum"),
        Average_Rating=("Rating", "mean"),
        Number_Of_Orders=("Product_ID", "count"),
        Maximum_Unit_Price=("Unit_Price", "max")
    )
    .reset_index()
)

print(product_features)
```

### Output Explanation

First, the Revenue column is calculated by multiplying Quantity and Unit_Price.

Then, records are grouped using Product_ID.

The aggregation operations calculate:

- Total_Quantity_Sold: Total units sold.
- Total_Revenue: Total revenue generated.
- Average_Rating: Average customer rating.
- Number_Of_Orders: Number of product records.
- Maximum_Unit_Price: Highest unit price.

The final dataset contains one row per product.

These features help businesses identify popular and profitable products.

## Topic 11 – Transaction-Level Aggregation


### Step 1: Understand the Concept

Transaction-Level Aggregation summarizes multiple product-line records belonging to the same transaction.

A single transaction can contain multiple products.

For example:

Transaction T101:
- Product P101: Quantity 2, Price ₹500
- Product P102: Quantity 1, Price ₹1000

Transaction Total = ₹2000

Transaction-level features can include:

- Total transaction amount.
- Total quantity of items.
- Number of different products.
- Average product price.

These features help understand the overall value and composition of each transaction.


### Step 2: Demonstrate the Concept


### Real-World Example

An online shopping company stores individual product records for every customer order.

The company wants to calculate the total value of each transaction.

Business Use:
- Calculate total order value.
- Analyze shopping basket size.
- Understand transaction composition.

AI/ML Use:
Transaction-level features can help predict order value, analyze shopping patterns, and support product recommendation systems.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Transaction_ID": ["T101", "T101", "T102", "T102", "T102", "T103"],
    "Product_ID": ["P101", "P102", "P101", "P103", "P104", "P102"],
    "Quantity": [2, 1, 1, 3, 2, 4],
    "Unit_Price": [500, 1000, 500, 200, 300, 1000]
})

# Calculate revenue for each product line
df["Line_Total"] = df["Quantity"] * df["Unit_Price"]

# Calculate transaction-level features
transaction_features = (
    df.groupby("Transaction_ID")
    .agg(
        Total_Transaction_Amount=("Line_Total", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Number_Of_Products=("Product_ID", "nunique"),
        Average_Product_Price=("Unit_Price", "mean")
    )
    .reset_index()
)

print(transaction_features)
```

### Output Explanation

First, Line_Total is calculated by multiplying Quantity and Unit_Price.

Then, records are grouped by Transaction_ID.

The aggregation operations calculate:

- Total_Transaction_Amount: Total value of the transaction.
- Total_Quantity: Total number of items purchased.
- Number_Of_Products: Number of unique products in the transaction.
- Average_Product_Price: Average unit price of the products.

The final dataset contains one row per transaction.

These features help businesses understand the value and composition of customer orders.

## Why Are Aggregation Features Useful?

Aggregation features are useful because they convert detailed records into meaningful summary information.

A dataset may contain thousands of individual transactions. Aggregation helps summarize those transactions at customer, product, or transaction level.

### 1. Understand Customer Behavior

Total spending, average order value, and purchase count help identify customer purchasing patterns.

### 2. Improve Business Decisions

Businesses can identify their best-selling products, highest-revenue regions, and most valuable customers.

### 3. Reduce Data Complexity

Instead of analyzing every transaction separately, we can create summarized datasets.

### 4. Identify Patterns

Mean, median, and standard deviation help understand typical values and variations in business data.

### 5. Improve Machine Learning Models

Aggregation features provide useful information to machine learning algorithms.

Examples:
- Customer churn prediction
- Customer segmentation
- Product recommendation
- Sales forecasting
- Fraud detection

### 6. Support Better Predictions

Historical customer spending, purchase frequency, and average order value can help models make better predictions.

### Important Consideration – Data Leakage

When creating aggregation features for machine learning, use only the data available before the prediction time.

Do not include future transactions when calculating historical customer or product features.

### Conclusion

Aggregation features transform raw records into meaningful numerical information.

They help businesses understand customers, products, transactions, and overall business performance.
