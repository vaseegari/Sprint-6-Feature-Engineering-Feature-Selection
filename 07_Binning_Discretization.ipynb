# Notebook 7 – Binning & Discretization

## Topic 1 – What is Binning?

### Step 1 – Understand the Concept

Binning is a data preprocessing technique used to convert continuous numerical values into a smaller number of groups called bins.

Instead of working with individual numerical values, we divide them into meaningful categories.

For example:

Age values:
18, 22, 25, 31, 38, 45, 52, 60

After binning:

- Young: 18–25
- Adult: 26–40
- Middle-Aged: 41–55
- Senior: 56 and above

Binning converts detailed numerical information into grouped categories.

It is also called Discretization.


### Step 2 – Demonstrate the Concept

Real-world example:

An online shopping company stores the ages of its customers.

Instead of analysing every individual age, the company groups customers into age categories.

| Customer | Age |
|---|---:|
| A | 22 |
| B | 29 |
| C | 35 |
| D | 48 |
| E | 62 |

After binning:

| Customer | Age | Age Group |
|---|---:|---|
| A | 22 | Young |
| B | 29 | Adult |
| C | 35 | Adult |
| D | 48 | Middle-Aged |
| E | 62 | Senior |

Business use:
- Customer segmentation.
- Marketing analysis.
- Understanding customer demographics.

AI/ML use:
Binning can simplify numerical features and help models identify patterns across groups.


### Step 3 – Implement the Concept

We will create a customer dataset and convert numerical Age values into categories using pandas.cut().

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E"],
    "Age": [22, 29, 35, 48, 62]
})

bins = [0, 25, 40, 55, 100]

labels = ["Young", "Adult", "Middle-Aged", "Senior"]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)

print(df)
```

### Output Explanation

The output contains the original Age column and a new Age_Group column.

The pd.cut() function divides the ages into predefined intervals.

For example:

- Age 22 belongs to Young.
- Age 29 belongs to Adult.
- Age 48 belongs to Middle-Aged.
- Age 62 belongs to Senior.

The numerical Age feature is still available, and the new categorical feature represents its age group.

## Topic 2 – Why Binning?

### Step 1 – Understand the Concept

Binning is used to simplify continuous numerical data by grouping similar values into categories.

Reasons for using binning:

1. Simplifies complex numerical data.
2. Helps identify patterns across groups.
3. Reduces the effect of small variations in numerical values.
4. Makes data easier to understand and interpret.
5. Helps create meaningful business categories.
6. Can help certain Machine Learning models learn useful patterns.

Example:

Instead of analysing customer incomes such as:

25000, 27000, 29000, 52000, 55000, 58000

We can group them into:

- Low Income
- Medium Income
- High Income



### Step 2 – Demonstrate the Concept

Real-world example:

A bank wants to analyse customers based on their annual income.

| Customer | Annual Income |
|---|---:|
| A | 24000 |
| B | 28000 |
| C | 45000 |
| D | 60000 |
| E | 90000 |

The bank groups customers into income categories.

Business use:
- Create customer segments.
- Design suitable financial products.
- Understand customer income distribution.

AI/ML use:
Grouped features may help models identify category-based patterns, especially when the relationship between a numerical variable and the target is not linear.

Note:
Binning does not always improve model performance. It should be tested before using it in a final model.



### Step 3 – Implement the Concept

We will group customer incomes into Low, Medium, and High categories using pandas.cut().

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E"],
    "Income": [24000, 28000, 45000, 60000, 90000]
})

bins = [0, 30000, 60000, 100000]

labels = ["Low Income", "Medium Income", "High Income"]

df["Income_Group"] = pd.cut(
    df["Income"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

print(df)
```

### Output Explanation

The output contains three income categories.

- Income up to 30000 is classified as Low Income.
- Income above 30000 up to 60000 is classified as Medium Income.
- Income above 60000 up to 100000 is classified as High Income.

Binning makes it easier to analyse customers according to their income levels.

## Topic 3 – Equal-Width Binning

### Step 1 – Understand the Concept

Equal-Width Binning divides the complete numerical range into bins having the same width.

Each bin covers an equal numerical interval.

Formula:

Bin Width = (Maximum Value - Minimum Value) / Number of Bins

Example:

Minimum Age = 10

Maximum Age = 50

Number of Bins = 4

Bin Width = (50 - 10) / 4

Bin Width = 10

The intervals will be:

- 10–20
- 20–30
- 30–40
- 40–50

Equal-width binning is useful when we want fixed numerical intervals.



### Step 2 – Demonstrate the Concept

Real-world example:

A company wants to group customers based on their ages.

Age values:

12, 18, 24, 30, 36, 42, 48

The company creates four equal-width bins.

| Age Range | Category |
|---|---|
| 10–20 | Group 1 |
| 20–30 | Group 2 |
| 30–40 | Group 3 |
| 40–50 | Group 4 |

Business use:
- Age-based customer segmentation.
- Employee age distribution.
- Customer demographic analysis.

AI/ML use:
Equal-width binning converts continuous values into fixed numerical intervals that models can use as categorical features.



### Step 3 – Implement the Concept

We will use pandas.cut() to create four equal-width bins.

The function can automatically calculate the intervals when we provide the number of bins.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [12, 18, 24, 30, 36, 42, 48]
})

df["Age_Bin"] = pd.cut(
    df["Age"],
    bins=4,
    labels=["Group 1", "Group 2", "Group 3", "Group 4"]
)

print(df)

print("\nBin intervals:")
print(pd.cut(df["Age"], bins=4).cat.categories)
```

### Output Explanation

The Age values are divided into four intervals with approximately equal widths.

Each interval covers the same numerical range.

The Age_Bin column displays the category assigned to each age.

Important:

Equal-width binning does not guarantee that every bin contains the same number of records.

Some bins may contain many values, while others may contain very few values.

## Topic 4 – Equal-Frequency Binning

### Step 1 – Understand the Concept

Equal-Frequency Binning divides numerical data into bins containing approximately the same number of observations.

It is also known as Equal-Depth Binning.

Unlike equal-width binning, the numerical width of each bin can be different.

Example:

Age values:

10, 12, 15, 20, 25, 30, 35, 40

Number of bins = 4

Each bin contains approximately 2 observations.

The groups are:

- Bin 1: 10, 12
- Bin 2: 15, 20
- Bin 3: 25, 30
- Bin 4: 35, 40

Equal-frequency binning is useful when the data distribution is uneven.


### Step 2 – Demonstrate the Concept

Real-world example:

An e-commerce company has customer spending data.

Most customers spend smaller amounts, while a few customers spend very large amounts.

The company wants to divide customers into four groups with approximately equal numbers of customers.

Business use:
- Customer segmentation.
- Spending-level analysis.
- Customer ranking.

AI/ML use:
Equal-frequency binning can reduce the problem of having too many observations in some bins and very few in others.

Note:
When multiple values are identical, it may not always be possible to create bins with exactly equal record counts.



### Step 3 – Implement the Concept

We will use pandas.qcut() to create four bins with approximately equal numbers of observations.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Spending": [100, 150, 200, 250, 500, 800, 1200, 2000]
})

df["Spending_Bin"] = pd.qcut(
    df["Spending"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

print(df)

print("\nNumber of customers in each bin:")
print(df["Spending_Bin"].value_counts().sort_index())
```

### Output Explanation

The pd.qcut() function divides the spending values into four quantile-based groups.

Each group contains approximately the same number of customers.

The numerical ranges are different because the spending values are not equally distributed.

The value_counts() function shows how many customers belong to each category.

## Topic 5 – Quantile Binning

### Step 1 – Understand the Concept

Quantile Binning divides numerical data based on quantiles.

A quantile is a value that separates sorted data into specified proportions.

Common quantiles:

- 25th percentile: First quartile (Q1)
- 50th percentile: Median (Q2)
- 75th percentile: Third quartile (Q3)
- 100th percentile: Maximum value

When we divide data into four quantile groups, each group contains approximately 25% of the observations.

Example:

Four quantile groups:

- Q1: Lowest 25%
- Q2: Next 25%
- Q3: Next 25%
- Q4: Highest 25%

Quantile binning is commonly implemented using pandas.qcut().



### Step 2 – Demonstrate the Concept

Real-world example:

A bank wants to classify customers according to their loan amounts.

It divides customers into four quantile groups.

| Category | Meaning |
|---|---|
| Q1 | Lowest loan amounts |
| Q2 | Lower-middle loan amounts |
| Q3 | Upper-middle loan amounts |
| Q4 | Highest loan amounts |

Business use:
- Loan portfolio analysis.
- Customer ranking.
- Financial risk analysis.

AI/ML use:
Quantile-based categories can help models identify differences between lower-value and higher-value observations.

Difference from equal-width binning:

Equal-width binning uses equal numerical ranges.

Quantile binning uses data distribution to determine the bin boundaries.



### Step 3 – Implement the Concept

We will create four quantile-based groups using pandas.qcut().

We will also calculate the quantile boundaries using quantile().

```python
import pandas as pd

df = pd.DataFrame({
    "Loan_Amount": [
        10000, 15000, 18000, 22000,
        30000, 35000, 45000, 50000,
        65000, 75000, 90000, 120000
    ]
})

df["Loan_Category"] = pd.qcut(
    df["Loan_Amount"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(df)

print("\nQuantile boundaries:")
print(df["Loan_Amount"].quantile([0, 0.25, 0.50, 0.75, 1.0]))
```

### Output Explanation

The Loan_Category column divides loan amounts into four quantile groups.

Q1 represents the lowest loan amounts.

Q4 represents the highest loan amounts.

The quantile() function displays the numerical boundaries used to understand the distribution.

Each group contains approximately 25% of the observations when there are no problematic repeated boundary values.

## Topic 6 – Domain-Based Binning

### Step 1 – Understand the Concept

Domain-Based Binning means creating categories using business knowledge, industry rules, or subject knowledge.

Instead of allowing an algorithm to decide the bin boundaries, we define meaningful ranges based on the problem.

Examples:

Healthcare:
- Low Risk
- Medium Risk
- High Risk

Banking:
- Low Income
- Medium Income
- High Income

Education:
- Poor Performance
- Average Performance
- Good Performance

Domain-based binning is useful when the categories have a clear real-world meaning.



### Step 2 – Demonstrate the Concept

Real-world example:

A bank wants to classify customers according to their credit scores.

Example categories:

| Credit Score | Risk Category |
|---|---|
| Below 580 | High Risk |
| 580–669 | Medium Risk |
| 670–739 | Low Risk |
| 740 and above | Very Low Risk |

Business use:
- Credit risk analysis.
- Loan eligibility assessment.
- Customer classification.

AI/ML use:
Domain-based categories can provide meaningful features based on established business rules.

Important:
The score boundaries above are illustrative. Actual credit-risk thresholds depend on the credit-scoring system and the institution's policy.



### Step 3 – Implement the Concept

We will create risk categories using predefined credit score ranges.

We will use pandas.cut() because the bin boundaries are defined by business rules.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E"],
    "Credit_Score": [520, 620, 700, 760, 820]
})

bins = [0, 580, 670, 740, 900]

labels = [
    "High Risk",
    "Medium Risk",
    "Low Risk",
    "Very Low Risk"
]

df["Risk_Category"] = pd.cut(
    df["Credit_Score"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

print(df)
```

### Output Explanation

The Risk_Category column is created using predefined credit score ranges.

Examples:

- Score 520: High Risk
- Score 620: Medium Risk
- Score 700: Low Risk
- Score 760: Very Low Risk

The categories are based on the example business rules rather than equal-width or equal-frequency calculations.

## Topic 7 – Continuous → Categorical

### Step 1 – Understand the Concept

Continuous data contains numerical values that can take many possible values within a range.

Examples:
- Age
- Income
- Temperature
- Height
- Weight

Categorical data represents groups or labels.

Examples:
- Young
- Adult
- Senior
- Low Income
- High Income

Continuous → Categorical conversion means converting continuous numerical values into categories using binning.

Example:

Original Age:

18, 25, 32, 45, 65

After conversion:

Young, Young, Adult, Middle-Aged, Senior

This process is also called discretization.



### Step 2 – Demonstrate the Concept

Real-world example:

A hospital records the body temperatures of patients.

Instead of analysing every individual temperature, it creates categories.

Illustrative categories:

- Below 36°C: Low
- 36°C to below 38°C: Normal
- 38°C and above: High

Business use:
- Patient monitoring.
- Medical data analysis.
- Reporting and classification.

AI/ML use:
Categorical versions of continuous features can help some models identify group-based patterns.

Important:
Binning a medical measurement can remove important details. Clinical decisions should use validated medical thresholds and the original measurements where appropriate.



### Step 3 – Implement the Concept

We will convert continuous Temperature values into categorical Temperature_Status values.

```python
import pandas as pd

df = pd.DataFrame({
    "Patient": ["A", "B", "C", "D", "E"],
    "Temperature": [35.5, 36.2, 37.0, 38.2, 39.0]
})

bins = [0, 36, 38, 45]

labels = ["Low", "Normal", "High"]

df["Temperature_Status"] = pd.cut(
    df["Temperature"],
    bins=bins,
    labels=labels,
    right=False
)

print(df)
```

### Output Explanation

The Temperature_Status column contains categorical values created from numerical temperatures.

Examples:

- 35.5°C: Low
- 36.2°C: Normal
- 37.0°C: Normal
- 38.2°C: High
- 39.0°C: High

The original Temperature column is preserved.

The new categorical feature provides a simplified representation of the temperature values.

## Advantages of Binning

1. Simplifies Data:
   Converts many numerical values into a smaller number of categories.

2. Easy Interpretation:
   Categories such as Low, Medium, and High are easier to understand.

3. Reduces Small Variations:
   Small differences between numerical values may be grouped into the same category.

4. Helps Identify Patterns:
   Makes it easier to compare groups and identify trends.

5. Supports Business Decisions:
   Helps with customer segmentation, risk classification, and income analysis.

6. Handles Nonlinear Relationships:
   Grouping a numerical feature may help certain models learn category-based patterns.

## Limitations of Binning

1. Information Loss:
   Individual numerical differences are lost when values are converted into categories.

2. Boundary Problems:
   Values close to a bin boundary may be assigned to different groups even when they are very similar.

3. Poor Bin Selection:
   Incorrect boundaries can create misleading categories.

4. Uneven Distribution:
   Equal-width bins may contain very different numbers of observations.

5. Increased Complexity:
   Too many categories can create additional features and make the model harder to understand.

6. Model Performance May Decrease:
   Binning can remove useful numerical information and may reduce prediction accuracy.

## Important Considerations

- Choose bin boundaries based on the problem.
- Understand the distribution before creating bins.
- Handle missing values separately.
- Check the number of records in each bin.
- Avoid creating too many categories.
- Compare model performance before and after binning.
- Learn bin boundaries from training data only and reuse them on validation and test data to avoid data leakage.

Conclusion:

Binning is useful for simplifying continuous numerical data and creating meaningful categories.

However, it can also cause information loss. Therefore, binning should be applied only when it provides useful information for the business problem or Machine Learning model.
