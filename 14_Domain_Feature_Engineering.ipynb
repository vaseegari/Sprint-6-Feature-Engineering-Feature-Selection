# Topic 3 – Feature Documentation and Business Justification


# Step 1 – Understand the Concept

Feature Documentation explains why a feature is created and how it should be calculated.

It helps data scientists, business analysts, and Machine Learning engineers understand the feature.

Each feature must have seven details:

1. Feature Name – Name of the new feature.
2. Source Columns – Original columns used.
3. Logic – Calculation or condition.
4. Business Meaning – What it represents in the business.
5. ML Relevance – How it may help the model.
6. Leakage Risk – Whether it may use unavailable or future information.
7. Retain/Remove Decision – Whether the feature should be included.

A feature should not be retained just because it can be calculated. It must have a clear purpose and be safe to use.



# Step 2 – Demonstrate the Concept

Example Feature: IsMonthToMonth

Otherwise, assign 0.

Business Meaning:
Identifies customers who have flexible monthly contracts.

ML Relevance:
Contract flexibility may be associated with customer churn.

Leakage Risk:
Low, provided the contract information is recorded before the prediction date.

Decision:
Retain.

This documentation explains both the business value and the safety of the feature.

# Topic 4 – Implementing Selected Domain Features


# Step 1 – Understand the Concept

After documenting possible features, we can implement the features that are marked for retention.

Feature implementation converts business rules into Python calculations.

However, feature creation must follow the documented logic.

For example:

TenureGroup:
Uses customer tenure to create customer duration categories.

TotalServices:
Counts subscribed telecom services.

IsMonthToMonth:
Converts contract type into a binary indicator.

We should not include the target column Churn in the feature calculations.



# Step 2 – Demonstrate the Concept

Suppose a customer has:

- Tenure = 8 months
- Contract = Month-to-month
- PhoneService = Yes
- InternetService = Fiber optic
- TechSupport = No

Possible engineered features:

- TenureGroup = New
- IsMonthToMonth = 1
- HasInternetService = 1
- HasTechnicalSupport = 0

These features convert raw customer information into meaningful business indicators.

The same business rules can be applied to other customer records.



# Step 3 – Implement the Concept

We will create the selected features using Pandas.

First, we will copy the original dataset to preserve the raw data.

Then we will create:

1. TenureGroup
2. IsMonthToMonth
3. TotalServices
4. HasInternetService
5. IsElectronicPayment
6. UsesPaperlessBilling
7. AverageChargesPerMonth
8. HighMonthlyCharges
9. HasFamily
10. HasTechnicalSupport
11. HasOnlineSecurity

CustomerID will remain available for record identification but will not be used as a model feature.

```python
# Keep the original dataset unchanged
domain_df = df.copy()

# Convert TotalCharges to numeric
domain_df["TotalCharges"] = pd.to_numeric(
    domain_df["TotalCharges"],
    errors="coerce"
)

# 1. Customer tenure group
domain_df["TenureGroup"] = np.select(
    [
        domain_df["tenure"] <= 12,
        domain_df["tenure"] <= 48
    ],
    [
        "New",
        "Medium-Term"
    ],
    default="Long-Term"
)

# 2. Month-to-month contract indicator
domain_df["IsMonthToMonth"] = (
    domain_df["Contract"] == "Month-to-month"
).astype(int)

# 3. Count subscribed services
service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

service_mapping = {
    "Yes": 1,
    "No": 0,
    "No internet service": 0,
    "No phone service": 0
}

service_data = domain_df[service_columns].replace(
    service_mapping
)

domain_df["TotalServices"] = service_data.sum(axis=1)

# 4. Internet service indicator
domain_df["HasInternetService"] = (
    domain_df["InternetService"] != "No"
).astype(int)

# 5. Electronic payment indicator
domain_df["IsElectronicPayment"] = (
    domain_df["PaymentMethod"] == "Electronic check"
).astype(int)

# 6. Paperless billing indicator
domain_df["UsesPaperlessBilling"] = (
    domain_df["PaperlessBilling"] == "Yes"
).astype(int)

# 7. Average charges per month
domain_df["AverageChargesPerMonth"] = np.where(
    domain_df["tenure"] > 0,
    domain_df["TotalCharges"] / domain_df["tenure"],
    np.nan
)

# 8. High monthly charges indicator
# For this demonstration, use the dataset median.
monthly_charge_threshold = domain_df["MonthlyCharges"].median()

domain_df["HighMonthlyCharges"] = (
    domain_df["MonthlyCharges"] > monthly_charge_threshold
).astype(int)

# 9. Family indicator
domain_df["HasFamily"] = (
    (domain_df["Partner"] == "Yes") |
    (domain_df["Dependents"] == "Yes")
).astype(int)

# 10. Technical support indicator
domain_df["HasTechnicalSupport"] = (
    domain_df["TechSupport"] == "Yes"
).astype(int)

# 11. Online security indicator
domain_df["HasOnlineSecurity"] = (
    domain_df["OnlineSecurity"] == "Yes"
).astype(int)

created_features = [
    "TenureGroup",
    "IsMonthToMonth",
    "TotalServices",
    "HasInternetService",
    "IsElectronicPayment",
    "UsesPaperlessBilling",
    "AverageChargesPerMonth",
    "HighMonthlyCharges",
    "HasFamily",
    "HasTechnicalSupport",
    "HasOnlineSecurity"
]

print("Number of Created Features:", len(created_features))

display(domain_df[created_features].head(10))
```

# Output Explanation

We successfully created 11 domain-based features.

These features describe:

- Customer relationship duration.
- Contract flexibility.
- Service adoption.
- Internet subscription.
- Payment preferences.
- Billing preferences.
- Historical spending.
- Family information.
- Support and security subscriptions.

The calculations follow the feature definitions documented earlier.

Important:
The monthly charge threshold in this demonstration uses the full dataset median for exploration only. Before building a predictive model, calculate the threshold using training data only and reuse that threshold on validation and test data.

# Topic 5 – Leakage Risk Assessment


# Step 1 – Understand the Concept

Leakage Risk Assessment means checking whether a feature contains information that would not be available when making a prediction.

A feature can be meaningful for the business but still be unsafe for Machine Learning.

Three useful risk categories:

1. Low Risk:
   Feature is available before the prediction time.

2. Medium Risk:
   Feature is useful, but its calculation depends on timing or data availability.

3. High Risk:
   Feature contains future information or information created after the target event.

Examples:

- Current Contract: Low risk.
- TotalCharges: Medium risk if it includes charges after the prediction date.
- Cancellation Date: High risk when predicting future churn.

We must define a prediction date before deciding whether a feature is safe.



# Step 2 – Demonstrate the Concept

Telecom churn prediction example:

Suppose we predict churn on 1 November.

Safe information:
- Contract type recorded on 1 November.
- Customer tenure on 1 November.
- Monthly charges on 1 November.

Unsafe information:
- Cancellation date recorded on 15 November.
- Termination reason recorded after cancellation.
- Future billing information recorded after 1 November.

Business understanding:

A company needs to identify customers who may leave before they actually leave.

Using information recorded after cancellation would make the model unrealistic.



# Step 3 – Implement the Concept

We will create a leakage assessment table.

The table will identify:

- Feature Name
- Leakage Risk
- Reason
- Recommended Action

This allows us to review each feature before using it in a Machine Learning model.

# Topic 6 – Final Retain or Remove Decisions


# Step 1 – Understand the Concept

Feature Retention means deciding which features should be included in the final Machine Learning dataset.

A feature can be retained when:

- It has a clear business meaning.
- It is relevant to the prediction problem.
- It is available at prediction time.
- It has acceptable data quality.
- It provides useful information.

A feature can be removed when:

- It is only an identifier.
- It contains future information.
- It reveals the target outcome.
- It duplicates unnecessary information.
- It does not provide useful predictive value.

Domain knowledge provides the initial decision. Model evaluation helps validate it.



# Step 2 – Demonstrate the Concept

Telecom example:

Retain:
- TenureGroup
- TotalServices
- IsMonthToMonth
- HasTechnicalSupport

Remove:
- CustomerID from model inputs.
- CancellationDate.
- TerminationReason.

Why?

The retained features describe customer behavior and subscriptions.

The removed features are identifiers or information that may only become available after the customer leaves.

This helps create a realistic prediction dataset.


# Step 3 – Implement the Concept

We will create a final decision table containing all documented features.

Then we will count how many features are marked for retention or removal.

```python
final_feature_decisions = feature_df[
    [
        "Feature Name",
        "Business Meaning",
        "ML Relevance",
        "Leakage Risk",
        "Retain/Remove Decision"
    ]
].copy()

display(final_feature_decisions)

print("\nFeature Decision Summary:")
print(
    final_feature_decisions["Retain/Remove Decision"]
    .value_counts()
)
```

# Output Explanation

These are initial domain decisions. Final feature selection should consider data quality, redundancy, validation performance, and business requirements.
