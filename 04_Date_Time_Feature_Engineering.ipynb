# 1. Year Feature Engineering

## Step 1 – Understand the Concept

**Definition:**

Year Feature Engineering is the process of extracting the year from a date column.

It helps identify the year in which an event or transaction happened.

**Example:**

Date: 2026-11-10

Extracted Year: 2026

**Formula:**

Year = Extract the four-digit year from a date.

**Why is it important?**

Year information helps identify yearly trends and patterns in business data.



## Step 2 – Demonstrate the Concept

**Real-World Example:**

An e-commerce company stores customer purchase dates.

The company wants to identify the year in which customers purchased products.

**Business Example:**

A company compares sales from 2024, 2025, and 2026 to understand yearly business growth.

**AI/ML Use Case:**

Year can be used as a feature in sales forecasting models to identify yearly trends.

**Sample Data:**

| Customer_ID | Purchase_Date |
|---|---|
| 101 | 2024-05-10 |
| 102 | 2025-08-15 |
| 103 | 2026-11-10 |

**Expected Output:**

| Customer_ID | Purchase_Year |
|---|---|
| 101 | 2024 |
| 102 | 2025 |
| 103 | 2026 |



## Step 3 – Implement the Concept

```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103],
    "Purchase_Date": [
        "2024-05-10",
        "2025-08-15",
        "2026-11-10"
    ]
}

df = pd.DataFrame(data)

df
```

### Convert Purchase_Date into Datetime

The Purchase_Date column is initially stored as a string.

We use pd.to_datetime() to convert it into datetime format.

Datetime conversion is required before extracting year information.

```python
df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"]
)

df.dtypes
```

### Extract Year

We use the `.dt.year` property to extract the year from the Purchase_Date column.

The extracted year is stored in a new column called Purchase_Year.

```python
df["Purchase_Year"] = df["Purchase_Date"].dt.year

df
```

### Output Explanation

- Customer 101 purchased the product in 2024.
- Customer 102 purchased the product in 2025.
- Customer 103 purchased the product in 2026.

The Purchase_Year column contains the extracted year from each purchase date.

**Conclusion:**

Year Feature Engineering converts raw date information into a useful numerical feature for business analysis and Machine Learning.

# 4. Day of Week Feature Engineering

## Step 1 – Understand the Concept

**Definition:**

Day of Week Feature Engineering is the process of identifying the weekday corresponding to a particular date.

Pandas represents weekdays using numbers:

- Monday = 0
- Tuesday = 1
- Wednesday = 2
- Thursday = 3
- Friday = 4
- Saturday = 5
- Sunday = 6

**Example:**

Date: 2026-11-10

Day of Week: Tuesday

Day Number: 1

**Why is it important?**

It helps identify customer behaviour on different weekdays and weekends.


## Step 2 – Demonstrate the Concept

**Real-World Example:**

An online shopping company wants to understand which weekdays customers prefer for purchasing products.

**Business Example:**

A supermarket compares weekday and weekend sales to plan staff schedules and stock availability.

**AI/ML Use Case:**

Day of Week can be used in sales forecasting models to identify weekly customer purchase patterns.

**Sample Data:**

| Customer_ID | Purchase_Date |
|---|---|
| 101 | 2026-11-09 |
| 102 | 2026-11-10 |
| 103 | 2026-11-14 |

**Expected Output:**

| Customer_ID | Purchase_Day |
|---|---|
| 101 | Monday |
| 102 | Tuesday |
| 103 | Saturday |




## Step 3 – Implement the Concept

```python
data = {
    "Customer_ID": [101, 102, 103],
    "Purchase_Date": [
        "2026-11-09",
        "2026-11-10",
        "2026-11-14"
    ]
}

df = pd.DataFrame(data)

df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"]
)

df["Day_Number"] = df["Purchase_Date"].dt.dayofweek

df["Day_Name"] = df["Purchase_Date"].dt.day_name()

df
```

### Output Explanation

- Customer 101 purchased on Monday.
- Customer 102 purchased on Tuesday.
- Customer 103 purchased on Saturday.

The Day_Number column contains the weekday number.

The Day_Name column contains the weekday name.

**Conclusion:**

Day of Week Feature Engineering helps businesses identify weekly customer behaviour and improve sales planning.

## Topic 5 – Week Number


### Step 1: Understand the Concept

Week Number represents the week of the year in which a particular date occurs.

It helps us identify when an event or transaction happened during the year.

For example:
- January 5 may belong to Week 1.
- February 15 may belong to Week 7.

Week numbers are useful for weekly sales reports, customer purchase analysis, and business forecasting.


### Step 2: Demonstrate the Concept

### Real-World Example

An online shopping company wants to analyze weekly sales.

Instead of analyzing every individual purchase date, the company groups transactions by week number.

Business Use:
- Weekly sales analysis
- Customer purchase patterns
- Weekly demand forecasting

AI/ML Use:
Week Number can help machine learning models identify weekly patterns in customer purchasing behavior.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Purchase_Date": [
        "2026-01-05",
        "2026-02-15",
        "2026-03-10",
        "2026-04-20"
    ]
})

df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])

df["Week_Number"] = df["Purchase_Date"].dt.isocalendar().week

print(df)
```

This feature helps businesses group and analyze transactions based on weeks.

## Topic 10 – Days Since Event


### Step 1: Understand the Concept

Days Since Event calculates the number of days between an event date and a reference date.

Formula:

Days Since Event = Reference Date - Event Date

For example:

Event Date: 2026-11-01
Reference Date: 2026-11-10

Days Since Event = 9 days

This feature helps measure how much time has passed since an event occurred.


### Step 2: Demonstrate the Concept


### Real-World Example

An e-commerce company wants to identify how many days have passed since a customer last contacted customer support.

Business Use:
- Days since last purchase
- Days since customer registration
- Days since last support request

AI/ML Use:
This feature can help predict customer activity and customer churn.

Important:
The reference date must not be earlier than the event date for this example.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Last_Event_Date": [
        "2026-11-01",
        "2026-11-05",
        "2026-11-08"
    ]
})

df["Last_Event_Date"] = pd.to_datetime(df["Last_Event_Date"])

reference_date = pd.Timestamp("2026-11-10")

df["Days_Since_Event"] = (
    reference_date - df["Last_Event_Date"]
).dt.days

print(df)
```

### Output Explanation

The Days_Since_Event column shows the number of days between the event date and the reference date.

For example:
- Customer 101: 9 days
- Customer 102: 5 days
- Customer 103: 2 days

This feature helps businesses understand how recently an event happened.

## Topic 11 – Time Difference


### Step 1: Understand the Concept

Time Difference calculates the duration between two timestamps.

Formula:

Time Difference = End Time - Start Time

The result can be measured in:
- Seconds
- Minutes
- Hours
- Days

It helps measure the time taken between two activities.



### Step 2: Demonstrate the Concept


### Real-World Example

An online delivery company wants to calculate how many hours a delivery takes.

Order Time: 2026-11-10 09:00
Delivery Time: 2026-11-10 13:30

Time Difference = 4.5 hours

Business Use:
- Delivery duration analysis
- Customer service response time
- Employee working hours

AI/ML Use:
Time Difference can help predict delivery delays and service completion times.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Time": [
        "2026-11-10 09:00:00",
        "2026-11-10 10:30:00",
        "2026-11-10 11:00:00"
    ],
    "Delivery_Time": [
        "2026-11-10 13:30:00",
        "2026-11-10 12:30:00",
        "2026-11-10 16:00:00"
    ]
})

df["Order_Time"] = pd.to_datetime(df["Order_Time"])
df["Delivery_Time"] = pd.to_datetime(df["Delivery_Time"])

df["Time_Difference_Hours"] = (
    df["Delivery_Time"] - df["Order_Time"]
).dt.total_seconds() / 3600

print(df)
```

### Output Explanation

The Time_Difference_Hours column represents the time taken between ordering and delivery.

For example:
- First order: 4.5 hours
- Second order: 2 hours
- Third order: 5 hours

Using total_seconds() allows us to calculate the duration accurately, including when timestamps cross midnight.

## Topic 12 – Customer Age


### Step 1: Understand the Concept

Customer Age calculates the age of a customer using their date of birth and a reference date.

Basic Formula:

Customer Age = Current Year - Birth Year

However, we must check whether the customer's birthday has already occurred in the reference year.

This feature converts date of birth into a numerical value.


### Step 2: Demonstrate the Concept


### Real-World Example

A banking company wants to analyze customers based on their age.

For example:
- Age 22: Young customer
- Age 35: Middle-age customer
- Age 60: Senior customer

Business Use:
- Customer segmentation
- Banking service recommendations
- Insurance analysis

AI/ML Use:
Customer Age can help predict customer preferences and financial product requirements.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Date_of_Birth": [
        "2000-05-15",
        "1995-12-20",
        "1988-03-10"
    ]
})

df["Date_of_Birth"] = pd.to_datetime(df["Date_of_Birth"])

reference_date = pd.Timestamp("2026-11-10")

df["Customer_Age"] = (
    reference_date.year
    - df["Date_of_Birth"].dt.year
    - (
        (reference_date.month < df["Date_of_Birth"].dt.month)
        | (
            (reference_date.month == df["Date_of_Birth"].dt.month)
            & (reference_date.day < df["Date_of_Birth"].dt.day)
        )
    ).astype(int)
)

print(df)
```

### Output Explanation

The Customer_Age column contains the calculated age of each customer.

The implementation checks whether the customer's birthday has occurred before the reference date.

This provides a more accurate age than subtracting only the birth year.

## Topic 13 – Customer Tenure


### Step 1: Understand the Concept

Customer Tenure measures how long a customer has been associated with a company.

Formula:

Customer Tenure = Reference Date - Joining Date

It can be measured in days, months, or years.

For example:

Joining Date: 2024-01-01
Reference Date: 2026-01-01

Customer Tenure = Approximately 2 years


### Step 2: Demonstrate the Concept


### Real-World Example

A subscription company wants to understand how long customers have been using its services.

Business Use:
- Customer loyalty analysis
- Customer retention
- Subscription management

AI/ML Use:
Customer Tenure can help predict customer churn and identify long-term customers.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Joining_Date": [
        "2024-01-01",
        "2025-06-15",
        "2026-01-10"
    ]
})

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

reference_date = pd.Timestamp("2026-11-10")

df["Tenure_Days"] = (
    reference_date - df["Joining_Date"]
).dt.days

df["Tenure_Years"] = df["Tenure_Days"] / 365.25

print(df)
```

### Output Explanation

The Tenure_Days column shows how many days each customer has been associated with the company.

The Tenure_Years column converts the tenure into approximate years.

365.25 is used as an approximate number of days per year, accounting for leap years over time.

## Topic 14 – Recency


### Step 1: Understand the Concept

Recency measures how recently a customer performed an activity, such as making a purchase.

Formula:

Recency = Reference Date - Last Purchase Date

A smaller recency value means the customer purchased recently.

A larger recency value means more time has passed since the last purchase.

Recency is an important part of RFM analysis.


### Step 2: Demonstrate the Concept


### Real-World Example

An online shopping company wants to identify customers who have not purchased products recently.

For example:
- Customer A: Last purchase 2 days ago
- Customer B: Last purchase 30 days ago
- Customer C: Last purchase 90 days ago

Business Use:
- Customer retention campaigns
- Promotional offers
- Customer segmentation

AI/ML Use:
Recency can help identify inactive customers and predict customer churn.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Last_Purchase_Date": [
        "2026-11-08",
        "2026-10-20",
        "2026-09-15"
    ]
})

df["Last_Purchase_Date"] = pd.to_datetime(
    df["Last_Purchase_Date"]
)

reference_date = pd.Timestamp("2026-11-10")

df["Recency_Days"] = (
    reference_date - df["Last_Purchase_Date"]
).dt.days

print(df)
```

### Output Explanation

The Recency_Days column represents the number of days since the customer's last purchase.

- Customer 101: 2 days
- Customer 102: 21 days
- Customer 103: 56 days

Customers with smaller recency values have purchased more recently.

This feature is useful for customer behavior analysis and churn prediction.

## Topic 15 – Frequency


###  Step 1: Understand the Concept

Frequency measures how many times a customer performs an activity within a specific period.

For example:
- Customer 101: 3 purchases
- Customer 102: 5 purchases
- Customer 103: 2 purchases

It helps identify how frequently customers interact with a business.

Frequency is another important part of RFM analysis.



### Step 2: Demonstrate the Concept


### Real-World Example

An e-commerce company wants to calculate how many purchases each customer makes.

Business Use:
- Identify frequent customers
- Customer loyalty programs
- Sales and marketing analysis

AI/ML Use:
Frequency can help predict customer loyalty, repeat purchases, and customer churn.

Important:
Frequency must be calculated using transactions available before the prediction date to avoid data leakage.


### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 101, 102, 102, 102, 103, 103, 103, 103],
    "Transaction_Date": [
        "2026-10-01",
        "2026-10-15",
        "2026-10-05",
        "2026-10-10",
        "2026-10-20",
        "2026-10-02",
        "2026-10-08",
        "2026-10-18",
        "2026-10-25"
    ]
})

df["Transaction_Date"] = pd.to_datetime(
    df["Transaction_Date"]
)

reference_date = pd.Timestamp("2026-11-01")

historical_data = df[
    df["Transaction_Date"] < reference_date
]

frequency = (
    historical_data.groupby("Customer_ID")
    .size()
    .reset_index(name="Purchase_Frequency")
)

print(frequency)
```

### Output Explanation

The Purchase_Frequency column shows the total number of transactions for each customer.

- Customer 101: 2 purchases
- Customer 102: 3 purchases
- Customer 103: 4 purchases

The groupby() function groups transactions by Customer_ID.

The size() function counts the number of transactions in each group.

This feature helps identify frequent customers.

## Topic 16 – Seasonality


### Step 1: Understand the Concept

Seasonality refers to patterns that repeat during specific periods of time.

These patterns may occur monthly, quarterly, or yearly.

Examples:
- Higher shopping activity during festival seasons
- More umbrella sales during rainy seasons
- Higher travel bookings during holidays

Seasonality helps identify repeated business patterns over time.



### Step 2: Demonstrate the Concept

### Real-World Example

An online shopping company wants to understand seasonal customer purchasing behavior.

For example:
- Summer: Higher demand for cooling products
- Monsoon: Higher demand for rain-related products
- Winter: Higher demand for winter clothing

Business Use:
- Seasonal sales planning
- Inventory management
- Marketing campaigns

AI/ML Use:
Seasonality features help time-series forecasting models predict future sales and customer demand.

Note:
Season definitions depend on the business location and industry.



### Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Purchase_Date": [
        "2026-01-15",
        "2026-04-15",
        "2026-07-15",
        "2026-10-15"
    ],
    "Sales": [12000, 18000, 15000, 25000]
})

df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])

def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

df["Season"] = df["Purchase_Date"].dt.month.apply(assign_season)

print(df)
```

### Output Explanation

The Season column identifies the season associated with each purchase date.

The function uses the month number to assign a season.

This example uses a simplified seasonal mapping for demonstration. Actual seasons should be defined based on the business location and domain.

Seasonality helps businesses identify repeated sales patterns and plan future demand.
